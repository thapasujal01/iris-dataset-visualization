```python
# CPSMA-4313-01: Data Processing Visualization
# Student: Sujal Thapa
# Instructor: Nicholas Jacob
# Project: Create Dataset - Student Loan Debt by State

import pandas as pd
import requests
import io

# (c) Loading necessary packages
# We use 'requests' to fetch the webpage and 'pandas' to parse HTML tables.
# 'BeautifulSoup' is often used for complex HTML, but pandas.read_html is
# more efficient for simple tables found on EducationData.org.

# (e) The site is referenced and loaded properly.
url = 'https://educationdata.org/student-loan-debt-by-income-level'
header = {
  "User-Agent": "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/50.0.2661.75 Safari/537.36",
  "X-Requested-With": "XMLHttpRequest"
}

r = requests.get(url, headers=header)

# (f) Discussing gathering data from raw HTML
# Using pd.read_html to find all <table> tags on the page.
tables = pd.read_html(io.StringIO(r.text))

# Based on EducationData.org structure, the "Income vs Student Debt State Table"
# is typically the 4th table on this specific page.
# Correction: The actual index for the desired table is 2, not 3.
df_income_debt = tables[2]

# (g) Data Cleaning
# Financial data often comes as strings with '$' and ','. We must clean these
# to make the data usable for calculations.
def clean_currency(value):
    if isinstance(value, str):
        return float(value.replace('$', '').replace(',', ''))
    return value

df_income_debt['Median Annual Income'] = df_income_debt['Median Annual Income'].apply(clean_currency)
df_income_debt['Average Student Debt'] = df_income_debt['Average Student Debt'].apply(clean_currency)

# Adding a calculated column to add value to our "new" dataset
df_income_debt['Debt_to_Income_Ratio'] = (df_income_debt['Average Student Debt'] / df_income_debt['Median Annual Income']).round(4)

# Display the first few rows
print("Preview of Cleaned Dataset:")
print(df_income_debt.head())

# (i) Saving the dataset
# This file can now be uploaded to your GitHub repository.
df_income_debt.to_csv('US_Student_Debt_By_State_2026.csv', index=False)
```
