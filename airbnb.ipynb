```python
import pandas as pd

# 1. Load the dataset from the GitHub URL
url = "https://raw.githubusercontent.com/nurfnick/Data_Viz/main/Data_Sets/AB_NYC_2019.csv"
df = pd.read_csv(url)

# 2. Check null counts before imputation (optional but helpful)
print("Nulls before:", df.isnull().sum())

# 3. Impute 'reviews_per_month' with 0.0
# Logic: If there are no reviews, the rate per month is naturally 0.
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

# 4. Impute 'name' with "Unknown"
# Logic: Provides a string placeholder for missing text descriptions.
df['name'] = df['name'].fillna("Unknown")

# 5. Verify the results
print("\nNulls after:")
print(df[['reviews_per_month', 'name']].isnull().sum())
```
