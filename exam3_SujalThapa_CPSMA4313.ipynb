```python
import pandas as pd
import re

# (a) Load the data [cite: 15, 16, 17]
url = "https://raw.githubusercontent.com/nurfnick/Data_Viz/refs/heads/main/Data_Sets/Activity_Dataset_V1.csv"
df = pd.read_csv(url)

# (b) Remove the unnamed index column [cite: 18]
df = df.drop(columns=['Unnamed: 0'])

# (c) Clean column names using Regex [cite: 19, 20, 21]
# This removes '(%)' and any trailing spaces
df.columns = [re.sub(r'\s*\(\%\)', '', col).strip() for col in df.columns]
```

```python
# (c) Clean column names to remove (%) using Regex
# Note: \s* matches any whitespace, \(%) matches the literal string [cite: 20]
df.columns = df.columns.str.replace(r'\s*\(%\)', '', regex=True)

# (d) Convert 'activity day' to datetime [cite: 22]
df['activity_day'] = pd.to_datetime(df['activity_day'])

# (e) Impute 'total steps' NaNs and convert to int
# Using the median is often an 'appropriate' number for activity data
df['total_steps'] = df['total_steps'].fillna(df['total_steps'].median()).astype(int)

# (f) Convert 'avg pace' (MM:SS) to float [cite: 24, 25]
def pace_to_float(pace_str):
    if pd.isna(pace_str):
        return 0.0
    minutes, seconds = map(int, str(pace_str).split(':'))
    return minutes + (seconds / 60)

df['avg_pace'] = df['avg_pace'].apply(pace_to_float)
```

```python
# (d) Convert the activity day column to datetime [cite: 22]
# Check your print(df.columns) output.
# If it shows 'activity_day', use that. If it's 'activity day', use that.
target_col = 'activity_day' if 'activity_day' in df.columns else 'activity day'

df[target_col] = pd.to_datetime(df[target_col])

# (i) Which day of the week and workout type has the max 'max_cadence' [cite: 28]
# Extract the day name
df['day_of_week'] = df[target_col].dt.day_name()

# Find the maximum
max_row = df.loc[df['max_cadence'].idxmax()]

print(f"The maximum cadence was on a {max_row['day_of_week']} during a {max_row['workout_type']} workout.")
```
